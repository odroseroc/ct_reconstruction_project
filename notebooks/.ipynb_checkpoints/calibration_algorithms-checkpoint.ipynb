# Calibration for image acquisition

This notebook explores techniques to calibrate the image acquisition. In particular, here we deal with producing master flat- and dark-fields and how to process images by making use of these fields.

The techniques explored here follow lectures in astronomy by Michael Richmond, available in his educational [repository](http://spiff.rit.edu/classes/), particularly his [lecture](http://spiff.rit.edu/classes/phys445/lectures/darkflat/darkflat.html) on dark subtraction and flatfielding. Some sample images used here have been taken from Richmond's repository and converted to a normalized .tif file, as all his images are in .fit format. This was done with the python script `fit_to_tif.py` included in the `imaging` directory of this project.

Ok, so what we need to get rid of the instrumental artifacts in the images is to get a master dark frame and a master flat frame, each of which is obtained from several images. We will require

✅ A function to import the dark/flat frames and store them in a list,

✅ A function to find the mean value of the images, -> This is straightforward, so no function was created

✅ A function to find the median value from all the images on a pixel-by-pixel basis,

✅ To construct the master dark from these values,

🔲 A function to normalize each flat frame to its mean value,

🔲 A function to scale every flat frame to a common mean level,

🔲 To subtract the dark frame from each image and to "divide" each image by the flat frame.

First we proceed to create the general functions that will allow us to generate the master frames.

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io
from pathlib import Path
from typing import Union

DATA_DIR = Path('../data/richmond/sep20_2003_tif')
files = ['dark4_001.tif','dark4_002.tif','dark4_003.tif','dark4_004.tif']

img = io.imread(DATA_DIR/files[0])
plt.imshow(img, cmap='grey')
```

```python
def import_images(files: list[str]) -> list[np.ndarray]:
    """
    Imports a list of image files and returns them as np.ndarray's
    
    Parameters
    ----------
    files : list[str]
        List of paths of the image files

    Returns
    -------
    list[np.ndarray]
        The images as arrays
    """

    return [io.imread(path) for path in files]

# Test here

fileList = [DATA_DIR/file for file in files]
frames = import_images(fileList)
plt.imshow(frames[2])
```

```python
def median_by_pixel(imgs: list[np.ndarray]) -> np.ndarray:
    """
    Determines the median value of a set of images, on a pixel by pixel basis.

    Parameters
    ----------
    imgs: list[np.ndarray]
        The images used to compute the median.

    Return
    ------
    np.ndarray
        An image whose pixels are the median values of the original images.
    """
    if not(all(im.shape==imgs[0].shape for im in imgs)):
        raise Exception('The images have different sizes.')

    original_shape = imgs[0].shape
    imgs = np.vstack([im.reshape((1,-1)) for im in imgs])
    return np.median(imgs, axis=0).reshape(original_shape)
```

```python
master_dark = median_by_pixel(frames)
plt.imshow(master_dark)
```

### Creation of the master dark frame

The master dark frame is simply the pixel-by-pixel median of all the dark frames given to the function. So, our function to produce the master dark frame will take the list of addresses of the dark frames and output a single tiff image containing the median-by-pixel of them. It would be convenient to receive the address of the output file as an optional argument. It would be also convenient to have a parameter that returns the master dark as an `np.ndarray`, for ease of use directly in other python scripts.

#### Comment:
I think the option to return the array is not required. The function can always return the array, but it will only be used if assigned to a variable. The function always exports the master dark, in any case.

```python
darkFrame_files = [Path(str(DATA_DIR)+"/dark4_%.3d.tif"%i) for i in range(1,11)]
flatFrame_files = [Path(str(DATA_DIR)+"/flatclear_%.3d.tif"%i) for i in range(1,11)]
```

```python
from collections.abc import Callable

def create_master_dark(input_files: list[Path],
                       output_file: Union[str, Path] = './masterdark.tif',
                       log_fn: Callable[[str],None] = print) -> np.ndarray:
    """
    Creates a master dark frame from a list of given images and exports it as
    an image.

    Parameters
    ----------
    input_files: list[Path]
        List of paths to the dark frames from which the master wil be created.
    output_file: str or Path, optional
        Path where the master dark frame is to be saved. If not given, it will
        be saved as masterdark.tif in the same directory as the original frames.
    log_fn: Callable, optional
        Function to show status messages. Must accept a string. By default it
        is the print function.

    Return
    ------
    np.ndarray
        Array containing the master dark frame.
    """
    
    darkFrames = import_images(input_files)
    output_file = Path(output_file)
    log_fn("Creating master dark frame from "+str(len(darkFrames))+" uploaded images")
    master_dark = median_by_pixel(darkFrames)
    io.imsave(output_file, master_dark, check_contrast=False)
    log_fn("Master dark frame saved in "+str(output_file))
    return master_dark

# Test here
master_dark = create_master_dark(darkFrame_files)
```

### Creation of the master flat frame

The master flat frame is created in three steps:

1. Subtract master dark frame from each flatfield image,
2. Obtain the mean of each image and normalize every pixel with respect to it -> Relative sensitivity of each pixel.
3. Scale each flatfield image to a common mean level.
4. Combine all the rescaled images into the master flat by taking the median pixel-by-pixel.

```python
from typing import Union, List, Tuple

def normalize_to_mean(imgs: Union[np.ndarray, List[np.ndarray]]) -> List[np.ndarray]:
    """
    Normalize one or more images by their mean intensity.

    Each image is divided by its mean pixel value. If the input is a single image,
    it is automatically wrapped into a list. If the mean of an image is zero,
    the image is returned unchanged.

    Parameters
    ----------
    imgs : np.ndarray or list of np.ndarray
        A single image or a list of images to normalize. Each image must be a NumPy array.

    Returns
    -------
    normalized_imgs : list of np.ndarray
        The normalized images, with each image divided by its mean.

    Notes
    -----
    If an image has a mean of zero, it is left unchanged to avoid division by zero.
    """
    if isinstance(imgs, np.ndarray):
        imgs = [imgs]
    means = [np.mean(img) for img in imgs]
    return [img/mean if mean != 0 else img for img, mean in zip(imgs, means)]

# Test here
norm_flats= normalize_to_mean(flatFrames)
print([np.mean(nf) for nf in norm_flats])
```

The images taken in the laboratory have different names, and it would be useful to be able to import them with a regex or a glob expression. Thus, we modify the import_images function to accept this kinds of arguments.

```python
from pathlib import Path
import matplotlib.pyplot as plt
from typing import Union, List
import numpy as np
from skimage import io
import re
import glob
from collections.abc  import Callable

DATA_DIR_STR = '../data/lab_images/flatframes/'

def import_images(paths: Union[List[str], str], log_fn: Callable = print) -> list[np.ndarray]:
    if isinstance(paths, str):
        paths = glob.glob(paths)
    paths = [Path(path) for path in paths]
    imgs = [io.imread(path) for path in paths]
    log_fn('Imported %d images'%len(paths))
    return [io.imread(path) for path in paths]

flatframes_30ms = import_images(DATA_DIR_STR+"flat30ms_*.tiff")
plt.imshow(flatframes_30ms[0], vmax=200)
```

Since the images taken in the lab are 8-bit tiffs, it is convenient to create a function that scales the resulting arrays to have only values between 0 and 255. This function should discard negative values arising from subtractions, such as when the master dark frame is subtracted from another image. We assume that the original image is also an 8-bit tiff, so that we can preserve the relative values between different images. This function should:

1. Read minimum and maximum value of the original image. The max value does not exceed 255.
2. If the minimum value is negative, trim all the negative values to 

```python
import sys
sys.path.append('..')
from collections.abc import Callable
from core.log_utils import no_op

def scale_to_mean(img: np.ndarray, mean_value: float,
                  img_mean: float = None,
                  log_fn: Callable[[str],None] = no_op) -> np.ndarray:
    """
    Scales the values of an image such that its mean is mean_value.
    Allows to parse the mean of the image as argument. This avoids repetitionif 
    the mean of this image has been previously calculated.
    """
    if img_mean == None:
        img_mean = np.mean(img)
    log_fn('')
    return img*(mean_value/img_mean)
```

The master flat is created first from the flat frames taken, by first scaling each image to a single mean value, and then taking the median-by-pixel of all of them. From that main-flat we can create the sensitivity map of each pixel, which is the image normalized to its mean. In this manner, all the pixels have a value between 0 and 1, and the sample image can be directly divided by this sensitivity map to adjust for the different sensitivities across the sensor.
